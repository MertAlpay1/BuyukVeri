```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Bronze") \
    .master("local[*]") \
    .config("spark.jars.packages","org.apache.spark:spark-sql-kafka-0-10_2.12:3.5.1,io.delta:delta-spark_2.12:3.1.0") \
    .config("spark.sql.extensions", "io.delta.sql.DeltaSparkSessionExtension") \
    .config("spark.sql.catalog.spark_catalog", "org.apache.spark.sql.delta.catalog.DeltaCatalog") \
    .getOrCreate()

print("Spark:", spark.version)
```

```python
from pyspark.sql.types import *
from pyspark.sql.functions import from_json, col

schema = StructType([
    StructField("timestamp", StringType()),
    StructField("user_id", StringType()),
    StructField("event_type", StringType()),
    StructField("related_id", StringType()),
    StructField("data", StructType([
    StructField("iso_code", StringType()),
    StructField("country", StringType()),
    StructField("year", DoubleType()),
    StructField("coal_prod_change_pct", DoubleType()),
    StructField("coal_prod_change_twh", DoubleType()),
    StructField("gas_prod_change_pct", DoubleType()),
    StructField("gas_prod_change_twh", DoubleType()),
    StructField("oil_prod_change_pct", DoubleType()),
    StructField("oil_prod_change_twh", DoubleType()),
    StructField("energy_cons_change_pct", DoubleType()),
    StructField("energy_cons_change_twh", DoubleType()),
    StructField("biofuel_share_elec", DoubleType()),
    StructField("biofuel_elec_per_capita", DoubleType()),
    StructField("biofuel_cons_change_pct", DoubleType()),
    StructField("biofuel_share_energy", DoubleType()),
    StructField("biofuel_cons_change_twh", DoubleType()),
    StructField("biofuel_consumption", DoubleType()),
    StructField("biofuel_cons_per_capita", DoubleType()),
    StructField("carbon_intensity_elec", DoubleType()),
    StructField("coal_share_elec", DoubleType()),
    StructField("coal_cons_change_pct", DoubleType()),
    StructField("coal_share_energy", DoubleType()),
    StructField("coal_cons_change_twh", DoubleType()),
    StructField("coal_consumption", DoubleType()),
    StructField("coal_elec_per_capita", DoubleType()),
    StructField("coal_cons_per_capita", DoubleType()),
    StructField("coal_production", DoubleType()),
    StructField("coal_prod_per_capita", DoubleType()),
    StructField("electricity_generation", DoubleType()),
    StructField("biofuel_electricity", DoubleType()),
    StructField("coal_electricity", DoubleType()),
    StructField("fossil_electricity", DoubleType()),
    StructField("gas_electricity", DoubleType()),
    StructField("hydro_electricity", DoubleType()),
    StructField("nuclear_electricity", DoubleType()),
    StructField("oil_electricity", DoubleType()),
    StructField("other_renewable_electricity", DoubleType()),
    StructField("other_renewable_exc_biofuel_electricity", DoubleType()),
    StructField("renewables_electricity", DoubleType()),
    StructField("solar_electricity", DoubleType()),
    StructField("wind_electricity", DoubleType()),
    StructField("energy_per_gdp", DoubleType()),
    StructField("energy_per_capita", DoubleType()),
    StructField("fossil_cons_change_pct", DoubleType()),
    StructField("fossil_share_energy", DoubleType()),
    StructField("fossil_cons_change_twh", DoubleType()),
    StructField("fossil_fuel_consumption", DoubleType()),
    StructField("fossil_energy_per_capita", DoubleType()),
    StructField("fossil_cons_per_capita", DoubleType()),
    StructField("fossil_share_elec", DoubleType()),
    StructField("gas_share_elec", DoubleType()),
    StructField("gas_cons_change_pct", DoubleType()),
    StructField("gas_share_energy", DoubleType()),
    StructField("gas_cons_change_twh", DoubleType()),
    StructField("gas_consumption", DoubleType()),
    StructField("gas_elec_per_capita", DoubleType()),
    StructField("gas_energy_per_capita", DoubleType()),
    StructField("gas_production", DoubleType()),
    StructField("gas_prod_per_capita", DoubleType()),
    StructField("hydro_share_elec", DoubleType()),
    StructField("hydro_cons_change_pct", DoubleType()),
    StructField("hydro_share_energy", DoubleType()),
    StructField("hydro_cons_change_twh", DoubleType()),
    StructField("hydro_consumption", DoubleType()),
    StructField("hydro_elec_per_capita", DoubleType()),
    StructField("hydro_energy_per_capita", DoubleType()),
    StructField("low_carbon_share_elec", DoubleType()),
    StructField("low_carbon_electricity", DoubleType()),
    StructField("low_carbon_elec_per_capita", DoubleType()),
    StructField("low_carbon_cons_change_pct", DoubleType()),
    StructField("low_carbon_share_energy", DoubleType()),
    StructField("low_carbon_cons_change_twh", DoubleType()),
    StructField("low_carbon_consumption", DoubleType()),
    StructField("low_carbon_energy_per_capita", DoubleType()),
    StructField("nuclear_share_elec", DoubleType()),
    StructField("nuclear_cons_change_pct", DoubleType()),
    StructField("nuclear_share_energy", DoubleType()),
    StructField("nuclear_cons_change_twh", DoubleType()),
    StructField("nuclear_consumption", DoubleType()),
    StructField("nuclear_elec_per_capita", DoubleType()),
    StructField("nuclear_energy_per_capita", DoubleType()),
    StructField("oil_share_elec", DoubleType()),
    StructField("oil_cons_change_pct", DoubleType()),
    StructField("oil_share_energy", DoubleType()),
    StructField("oil_cons_change_twh", DoubleType()),
    StructField("oil_consumption", DoubleType()),
    StructField("oil_elec_per_capita", DoubleType()),
    StructField("oil_energy_per_capita", DoubleType()),
    StructField("oil_production", DoubleType()),
    StructField("oil_prod_per_capita", DoubleType()),
    StructField("other_renewables_elec_per_capita", DoubleType()),
    StructField("other_renewables_share_elec", DoubleType()),
    StructField("other_renewables_cons_change_pct", DoubleType()),
    StructField("other_renewables_share_energy", DoubleType()),
    StructField("other_renewables_cons_change_twh", DoubleType()),
    StructField("other_renewable_consumption", DoubleType()),
    StructField("other_renewables_energy_per_capita", DoubleType()),
    StructField("per_capita_electricity", DoubleType()),
    StructField("population", DoubleType()),
    StructField("primary_energy_consumption", DoubleType()),
    StructField("renewables_elec_per_capita", DoubleType()),
    StructField("renewables_share_elec", DoubleType()),
    StructField("renewables_cons_change_pct", DoubleType()),
    StructField("renewables_share_energy", DoubleType()),
    StructField("renewables_cons_change_twh", DoubleType()),
    StructField("renewables_consumption", DoubleType()),
    StructField("renewables_energy_per_capita", DoubleType()),
    StructField("solar_share_elec", DoubleType()),
    StructField("solar_cons_change_pct", DoubleType()),
    StructField("solar_share_energy", DoubleType()),
    StructField("solar_cons_change_twh", DoubleType()),
    StructField("solar_consumption", DoubleType()),
    StructField("solar_elec_per_capita", DoubleType()),
    StructField("solar_energy_per_capita", DoubleType()),
    StructField("gdp", DoubleType()),
    StructField("wind_share_elec", DoubleType()),
    StructField("wind_cons_change_pct", DoubleType()),
    StructField("wind_share_energy", DoubleType()),
    StructField("wind_cons_change_twh", DoubleType()),
    StructField("wind_consumption", DoubleType()),
    StructField("wind_elec_per_capita", DoubleType()),
    StructField("wind_energy_per_capita", DoubleType()),

    ]))
])

#Test için bu değer kullanıldı değiştirilebilir

df_raw = spark.readStream \
    .format("kafka") \
    .option("kafka.bootstrap.servers", "kafka:29092") \
    .option("subscribe", "world_energy_consumption") \
    .option("startingOffsets", "earliest") \
    .load()

df_parsed = df_raw.select(
    from_json(col("value").cast("string"), schema).alias("msg")
).select("msg.*")




print("Stream hazır")
```

```python
df_final = df_parsed.select(
    "timestamp",
    "user_id",
    "event_type",
    "related_id",
    "data.*"
)

bronze_query = df_final.writeStream \
    .format("delta") \
    .option("checkpointLocation", "/home/jovyan/data/checkpoints/bronze") \
    .outputMode("append") \
    .start("/home/jovyan/data/bronze")

print("Veri toplanıyor")
bronze_query.awaitTermination(30)
print("Veri toplandı")
```
